# **<font color='#3250a8'>Data Processing</font>**

---



Preview dataset structure


```python
print("Dataset Overview:")
df.info()
```

The Age column had a mix of numbers and spelled-out words like "twenty five", so it was standardized into proper numeric values.

Data Cleaning Step 1: Fixing Data Type, assign clean value


```python
print(df['Age'].unique())
```

Convert Age column from Object(string) to float


```python
def clean_age(val):
  if pd.isna(val):
    return val
  elif val == 'twenty five':
    return 25
  else:
    return float(val)

df['Age'] = df['Age'].apply(clean_age)
print('Age Column Succesfully Cleaned!\n ')
print(df['Age'].unique())
print('\n',df.dtypes)
```

Age is now clean float64 column with no text values remaining.

Data Cleaning Step 2: Fix Data Types


```python
print('Data Cleaning: Step 2')
```

Order_Date contained three different date formats, so it was converted into a single consistent datetime format using pandas' mixed-format parser.

Convert Date column from Object to Datetime


```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'],format = 'mixed')
print(df.dtypes)
print(df['Order_Date'].head())
#df['Order_Date'].isna().sum()
```

All 1000 dates parsed successfully with zero missing values, confirming the conversion worked.

Missing values were treated differently depending on the column - numbers were filled using sensible estimates, while text columns were marked as 'Unknown' instead of guessing

Data Cleaning Step 3: Handling Missing values


Handling missing values in Price


Missing prices were estimated using the median price within the same category, since prices vary a lot across categories like Electronics and Beauty.

```python
df['Price'] = df.groupby('Category')['Price'].transform(lambda x: x.fillna(x.median()))
print(df['Price'].isna().sum())
```

Handling missing values in Gender,City and Payment method



Gender, City, and Payment_Method had missing values too, but since these are just categories (not numbers), I couldn't estimate them like Price. So I filled the gaps with "Unknown" instead of guessing.

```python
df['Gender'] = df['Gender'].fillna('Unknown')
df['City'] = df['City'].fillna('Unknown')
df['Payment_Method'] = df['Payment_Method'].fillna('Unknown').str.strip().str.title()

print(df[['Gender','City','Payment_Method']].isna().sum())
```

Handling Missing values in Rating


Rating was left as missing (NaN) instead of filled, because a missing rating likely means the customer never left one — filling it with a number would misrepresent that as real feedback.

```python
print(df['Rating'].isna().sum())
```

Removing Duplicates


The dataset was checked for duplicate rows, which would cause an order to be counted more than once.

```python
before = df.shape[0]
df = df.drop_duplicates()
after = df.shape[0]
print(f'Rows before: {before}')
print(f'Rows after: {after}')
print(f'Duplicates removed: {before - after}')
```

No exact duplicates were found — all 1000 rows are unique, so no rows needed to be removed.

Creating a Derived Column: Total_Sale

The dataset had Price and Quantity as separate columns, but no single column showing the actual amount spent per order.

```python
df['Total_Sale'] = df['Price']*df['Quantity']
print(df[['Price','Quantity','Total_Sale']].head())
print(df['Price'].isna().sum())
```

A new column, Total_Sale, was created by multiplying Price and Quantity together, giving the real order value to use in later analysis.

Creating a Derived Column: Order_Month

Order_Date contained the full date,
but there was no easy way to group orders by month.

```python
df['Order_Month'] = df['Order_Date'].dt.month_name().str[:3]
print(df[['Order_Date','Order_Month']].head())
```

A new column, Order_Month, was created by extracting just the month from Order_Date, making it possible to analyze trends over time.

# **<font color='#8756b8'>Analysis Questions</font>**

---




Question 1: Which category generates the most revenue?

This question checks which product category brings in the most money overall, helping identify where the business is performing best.

```python
top_category = revenue_by_category.idxmax()
top_value = revenue_by_category.max()
print(f"The category with the highest revenue is {top_category}, generating {top_value:.0f} in total sales.")
```

Clothing generates the highest revenue overall (₹47,588,240), making it the strongest-performing category in the dataset.

Question 2: Which city has the most orders?

This question identifies which city generates the most order volume, which is useful for understanding where the biggest customer base is.

```python
top_city = df['City'].value_counts().idxmax()
top_city_count = df['City'].value_counts().max()
print(f"{top_city} has the most orders, with {top_city_count} total orders.")
```

Lahore has the highest order volume with 191 orders, suggesting it's the largest or most active customer base among the cities in this dataset.

Question 3: What percentage of customers paid by each method?

This question shows how customers prefer to pay, which can guide decisions like which payment options to prioritize or promote.

```python
payment_pct = df['Payment_Method'].value_counts(normalize=True) * 100
print(payment_pct.round(1))
```

Cash (33.4%) and Card (32.9%) are used almost equally, with COD trailing behind at 17.1% — there's no single dominant payment method, and 16.6% of orders have missing payment data.

Question 4: What is the average order value?

This question gives a single overall number representing how much a typical customer spends per order, useful as a quick benchmark.

```python
avg_order_value = df['Total_Sale'].mean()
print(f"The average order value is {avg_order_value:.2f}")
```

The average customer spends ₹164,192.60 per order, giving a useful benchmark for what a "typical" transaction looks like.
